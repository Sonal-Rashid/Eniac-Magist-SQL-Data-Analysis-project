```python
# uploading cleaned datasets from local
# orders
orders_qa_original = pd.read_csv('./data/orders_qa.csv', parse_dates=['created_date'])

# orderlines
orderlines_qa_original = pd.read_csv('./data/orderlines_qa.csv', parse_dates=['date'])

# products
products_brands_cg_original = pd.read_csv('./data/products_brands_cg.csv')
```

Creating a copy of the datasets. This ensures that our changes won't effect original DataFrames.

```python
orders_an = orders_qa_original.copy()
orderlines_an = orderlines_qa_original.copy()
products_brands_an = products_brands_cg_original.copy()
```

first let us find the total products that we have after the cleaning. We are also finding no of products that are discounted in this. We should find the same numbers in each category

```python
orderlines_products = orderlines_an.merge(products_brands_an, on='sku')
```

```python
orderlines_products['discount'] = orderlines_products['price']-orderlines_products['unit_price']
orderlines_products
```

```python
orderlines_products.sample(5)
```

```python
orderlines_products['discount'].describe()
```

```python
negative_order_id = orderlines_products.loc[(orderlines_products['discount'] < 0), 'order_id']
```

```python
orders_an = orders_an.loc[~orders_an['order_id'].isin(negative_order_id)]
```

```python
orderlines_an = orderlines_an.loc[~orderlines_an['order_id'].isin(negative_order_id)]
```

```python
orderlines_products = orderlines_products.loc[~orderlines_products['order_id'].isin(negative_order_id)]
```

```python
orders_an['order_id'].nunique(), orderlines_an['order_id'].nunique()
```

```python
orderlines_products['short'].isna().sum()
```

```python
orderlines_products = orderlines_products[['order_id', 'sku', 'product_quantity', 'date', 'unit_price', 'price', 'discount', 'category', 'brands']]
```

```python
discount_mask = ~(orderlines_products['discount'] == 0)
orderlines_products.loc[discount_mask, 'product_quantity'].sum()
```

```python
# percentage of products discounted
((orderlines_products.loc[discount_mask, 'product_quantity'].sum() / orderlines_products['product_quantity'].sum()) * 100).round(2)
```
