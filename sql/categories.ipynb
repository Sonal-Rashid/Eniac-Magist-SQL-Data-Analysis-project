We have got 783 rows of products in type (1298), which are open box products and it contains products from various categories. So we need to make use of regex to categorize it

```python
import re
def category_from_text(row):
    text = f"{row.get('name', '')}{row.get('desc', '')}".lower()

    if re.search(r'case|protective|screen protector|bag|sleeve|cover|watch strap',text):
        return 'cases & protection'
    elif re.search(r'storage|memory|flash storage|storage upgrade|hard drive|hard disk|usb|ssd|pendrive|raid|server|synology',text):
        return 'storage'
    elif re.search(r'adapter|cable|charger|powerbank|hub|station|dock|wireless|connector|ethernet|converter|hdmi',text):
        return 'power, cables & connectivity'
    elif re.search(r'headphone|earphone|headset|mouse|keyboard|drawing|stylus|pointer|speaker|keypad|hotspot|graphic',text):
        return 'input & audio devices'
    elif re.search(r'software|service|repair service',text):
        return 'software & repair sevices'
    elif re.search(r'ifixit|repair|kit|replacement|expansion',text):
        return 'repair tools & parts'
    elif re.search(r'stand|support|armband|accessories|mount|drawing|device care|remote|strips|color manager|mat',text):
        return 'accessories & miscellaneous'
    elif re.search(r'imac|macbook|mac mini|monitor|desktop',text):
        return 'computers & laptops'   
    elif re.search(r'watch|iphone|ipod|ipad|smart|sensor|camera|robot|amplifier|bulb',text):
        return 'mobile, wearables & smart devices'
    
```

```python
# calling the function
type_mask = products_cg['type'] == '1298'

products_cg.loc[type_mask, 'category'] = (
    products_cg.loc[type_mask]
    .apply(category_from_text, axis=1)
)
```
